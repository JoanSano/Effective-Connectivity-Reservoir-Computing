# Systems in Table 1 in Ning & Rathi, 2018 

### We start by generating the Vector Autoregressive processes

```python
import numpy as np
import os

# System 1 from Ning & Rathi, 2018
A_real = np.array([
    [.3,.3,0],
    [.3,.3,.3],
    [0,0,.3]
]) 
np.savetxt(os.path.join("Datasets/3-VAR_Ning-Rathi_1","Networks", "network.tsv"), A_real, delimiter='\t')
for s in range(1,21):
    T = 1080
    ts = np.zeros((A_real.shape[0],T))
    ts[:,0] = np.random.randn(A_real.shape[0])
    ts[:,1] = np.random.randn(A_real.shape[0])
    for i in range(2,T):
        ts[:,i] = ts[:,i-1] @ A_real + np.random.randn(A_real.shape[0])
    ts = ts.T
    to_save = np.zeros((T,4))
    to_save[:,0] = ts[:,0] * np.nan
    to_save[:,1:] = ts
    np.savetxt(os.path.join("Datasets/3-VAR_Ning-Rathi_1","Timeseries", f"sub-{s}_VAR_TS.txt"), to_save, delimiter='\t')

# System 2 from Ning & Rathi, 2018
A_real = np.array([
    [.3,.3,0],
    [.3,.3,.3],
    [.3,0,.3]
]) 
np.savetxt(os.path.join("Datasets/3-VAR_Ning-Rathi_2","Networks", "network.tsv"), A_real, delimiter='\t')
for s in range(1,21):
    T = 1080
    ts = np.zeros((A_real.shape[0],T))
    ts[:,0] = np.random.randn(A_real.shape[0])
    ts[:,1] = np.random.randn(A_real.shape[0])
    for i in range(2,T):
        ts[:,i] = ts[:,i-1] @ A_real + np.random.randn(A_real.shape[0])
    ts = ts.T
    to_save = np.zeros((T,4))
    to_save[:,0] = ts[:,0] * np.nan
    to_save[:,1:] = ts
    np.savetxt(os.path.join("Datasets/3-VAR_Ning-Rathi_2","Timeseries", f"sub-{s}_VAR_TS.txt"), to_save, delimiter='\t')
```

### Spectral pairwise conditional granger causality

With this method we can study the results in the frequency domain, as well as the granger measures compared to permutations.

```python
import glob
import numpy as np
import matplotlib.pylab as plt

from analysis.ground_truths.var_models import var

system = 2

# Time averaged measures
VAR_spwcgc = var(f"Results-test_3-VAR_Ning-Rathi-{system}_Method-spwCGC")
binary_networks = VAR_spwcgc.get_networks_Adjacency(weighted=False, bidirectional=False, save_in=False)[0]
weighted_networks = VAR_spwcgc.get_networks_Adjacency(weighted=True, bidirectional=False, save_in=False)[0]
F = np.zeros(weighted_networks.shape) * np.nan
for i in range(weighted_networks.shape[1]):
    for j in range(i+1, weighted_networks.shape[1]):
        score = VAR_spwcgc.get_interaction_scores(i+1,j+1)[1]
        F[:,i,j] = score.i2j[:,0]
        F[:,j,i] = score.j2i[:,0]

# Additionally, we can obtain the frequency measures as well
frequency_measure = [np.load(f,allow_pickle=True) for f in glob.glob(f"Results-test_3-VAR_Ning-Rathi-{system}_Method-spwCGC/*/Numerical/frequency-measure.npy")]
frequency_measure_surrogates = [np.load(f,allow_pickle=True) for f in glob.glob(f"Results-test_3-VAR_Ning-Rathi-{system}_Method-spwCGC/*/Numerical/frequency-measure_surrogates.npy")]
averaged_measure_surrogates = [np.load(f,allow_pickle=True) for f in glob.glob(f"Results-test_3-VAR_Ning-Rathi-{system}_Method-spwCGC/*/Numerical/averaged-measure_surrogates.npy")]
```

```python
size = 3
subject = 12
f, x_axis = frequency_measure[subject][()]['f'], frequency_measure[subject][()]['frequency']
f_surrs, x_surrs = frequency_measure_surrogates[12][()]['f'], frequency_measure_surrogates[12][()]['frequency']

fig, axes = plt.subplots(size, size, figsize=(size*size+1,size*size+1))
for i in range(size):
    for j in range(size):
        axes[i,j].set_title("|"+r'$f_{\theta}$'+f"({i+1},{j+1})|")
        axes[i,j].spines[["top","right"]].set_visible(False)
        axes[i,j].set_xlim([0,x_axis[-1]])
        if i!= size-1:
            axes[i,j].set_xticks([])
        else:
            axes[i,j].set_xlabel(r'$\theta$')
        
        # Surrogates
        for s in range(f_surrs.shape[0]):
            axes[i,j].plot(x_surrs[s], f_surrs[s][i,j], linewidth=1.5, color='gray', alpha=0.3)
        
        # Real value
        axes[i,j].plot(x_axis, f[i,j], linewidth=1.5, color='black')
```

```python
F_surrs = averaged_measure_surrogates[subject]

fig, axes = plt.subplots(size, size, figsize=(size*size+1,size*size+1))
for i in range(size):
    for j in range(size):
        if i!=j:
            axes[i,j].set_title(r'$F_{cGCM}$'+f"({i+1},{j+1})")
            axes[i,j].spines[["top","right"]].set_visible(False)
            y, x, bar = axes[i,j].hist(F_surrs[:,i,j], bins=50, color='gray', alpha=0.5)
            p_val = 1-weighted_networks[subject,i,j] # For spwcgc, the pvalue is obtained via permutations
            axes[i,j].vlines(F[subject,i,j], 0, 10, linewidth=2, color='black', label=f"p = {round(p_val,3)}")
            axes[i,j].legend(frameon=False)

        else:
            axes[i,j].spines[["top","right","bottom","left"]].set_visible(False)
            axes[i,j].set_xticks([])
            axes[i,j].set_yticks([])
            
```

### Method comparisson

We can obtain AUROC and DET curves for each method to study their performance. We can constrain or not to the known structure. The consensus in these types os systems is not to constrain to the structure, so we won't. Later we will redoo the analysis constraining to the known structure.

```python
import numpy as np
import matplotlib.pylab as plt
from analysis.ground_truths.var_models import var
from analysis.ground_truths.utils import Metrics

system = 2

# Load the rest of the results
VAR_pwGC = var(f"Results-test_3-VAR_Ning-Rathi-{system}_Method-pwGC")
N_vars = len(VAR_pwGC.subjects)
w_pwGC = VAR_pwGC.get_networks_Adjacency(weighted=True, bidirectional=False)[-1]
b_pwGC = VAR_pwGC.get_networks_Adjacency(weighted=False, bidirectional=False)[-1]
gt_pwGC = VAR_pwGC.binary_gt - np.eye(b_pwGC.shape[1])
metrics_pwGC = Metrics(ground_truth=np.tile(gt_pwGC.flatten(),N_vars), weighted_predictions=w_pwGC.flatten(), binary_predictions=b_pwGC.flatten())
auc_pwGC = metrics_pwGC.auc(n_bootstrap=None, level=0.95)
sens_pwGC, spec_pwGC, ppv_pwGC, npv_pwGC = metrics_pwGC.compute_confusion_matrix()

VAR_pwCGC = var(f"Results-test_3-VAR_Ning-Rathi-{system}_Method-pwCGC")
w_pwCGC = VAR_pwCGC.get_networks_Adjacency(weighted=True, bidirectional=False)[0]
b_pwCGC = VAR_pwCGC.get_networks_Adjacency(weighted=False, bidirectional=False)[0]
gt_pwCGC = VAR_pwCGC.binary_gt - np.eye(b_pwCGC.shape[1])
metrics_pwCGC = Metrics(ground_truth=np.tile(gt_pwCGC.flatten(),N_vars), weighted_predictions=w_pwCGC.flatten(), binary_predictions=b_pwCGC.flatten())
auc_pwCGC = metrics_pwCGC.auc(n_bootstrap=None, level=0.05)
sens_pwCGC, spec_pwCGC, ppv_pwCGC, npv_pwCGC = metrics_pwCGC.compute_confusion_matrix()

VAR_spwCGC = var(f"Results-test_3-VAR_Ning-Rathi-{system}_Method-spwCGC")
w_spwCGC = VAR_spwCGC.get_networks_Adjacency(weighted=True, bidirectional=False)[0]
b_spwCGC = VAR_spwCGC.get_networks_Adjacency(weighted=False, bidirectional=False)[0]
gt_spwCGC = VAR_spwCGC.binary_gt - np.eye(b_spwCGC.shape[1])
metrics_spwCGC = Metrics(ground_truth=np.tile(gt_spwCGC.flatten(),N_vars), weighted_predictions=w_spwCGC.flatten(), binary_predictions=b_spwCGC.flatten())
auc_spwCGC = metrics_spwCGC.auc(n_bootstrap=None, level=0.05)
sens_spwCGC, spec_spwCGC, ppv_spwCGC, npv_spwCGC = metrics_spwCGC.compute_confusion_matrix()

print("pwGC")
print("-----------------")
print(f"AUC = {auc_pwGC}")
print(f"Sensitivity = {sens_pwGC}")
print(f"Specificity = {spec_pwGC}")
print(f"PPV = {ppv_pwGC}")
print(f"NPV = {npv_pwGC}")
print("+++++++++++++++++")
print("pwCGC")
print("-----------------")
print(f"AUC = {auc_pwCGC}")
print(f"Sensitivity = {sens_pwCGC}")
print(f"Specificity = {spec_pwCGC}")
print(f"PPV = {ppv_pwCGC}")
print(f"NPV = {npv_pwCGC}")
print("+++++++++++++++++")
print("spwCGC")
print("-----------------")
print(f"AUC = {auc_spwCGC}")
print(f"Sensitivity = {sens_spwCGC}")
print(f"Specificity = {spec_spwCGC}")
print(f"PPV = {ppv_spwCGC}")
print(f"NPV = {npv_spwCGC}")
print("+++++++++++++++++")

thresholds = [.1, .5, .8, .9, .95]
A_pwGC = np.zeros((len(thresholds),3,3))
A_pwCGC = np.zeros((len(thresholds),3,3))
A_spwCGC = np.zeros((len(thresholds),3,3))

fig, ax = plt.subplots(3,len(thresholds),figsize=(16,11))
for i, t in enumerate(thresholds):
    _, A_pwGC[i] = VAR_pwGC.group_consistency(lag=-1, threshold=t, make_nan=True, make_binary=True)
    _, A_pwCGC[i] = VAR_pwCGC.group_consistency(lag=0, threshold=t, make_nan=True, make_binary=True)
    _, A_spwCGC[i] = VAR_spwCGC.group_consistency(lag=0, threshold=t, make_nan=True, make_binary=True)

    ax[0,i].imshow(A_pwGC[i], cmap="Reds", vmin=0, vmax=1)
    ax[1,i].imshow(A_pwCGC[i], cmap="Blues", vmin=0, vmax=1)
    ax[2,i].imshow(A_spwCGC[i], cmap="Greens", vmin=0, vmax=1)

    ax[0,i].set_xticks([]), ax[0,i].set_yticks([])
    ax[1,i].set_xticks([]), ax[1,i].set_yticks([])
    ax[2,i].set_xticks([]), ax[2,i].set_yticks([])

    ax[0,i].set_title(f"Consistency threshold = {t}", fontsize=10, fontweight='bold')

    if i == 0:
        ax[0,i].set_ylabel("Method --> pwGC")
        ax[1,i].set_ylabel("Method --> pwCGC")
        ax[2,i].set_ylabel("Method --> spwCGC")
```
